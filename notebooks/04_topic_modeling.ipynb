# Topic Modeling: BERTopic on Captions & Justifications
Discovers dominant narrative themes and their cultural distribution.

```python
import sys; sys.path.insert(0, '..')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.analysis.data_loading import load_annotations
from src.analysis.topic_modeling import fit_bertopic, topic_culture_proportions, topic_sentiment_composition, save_topic_info
sns.set_theme(style='white')
```

```python
df = load_annotations(model_names=['qwen_vl'])
cap_embs = np.load('../outputs/analysis/caption_embs_qwen_vl.npy')

print(f'Fitting BERTopic on {len(df)} captions...')
model, topics, probs = fit_bertopic(df['caption'].fillna('').tolist(), cap_embs)
df['caption_topic'] = topics
print(f'Topics found: {len(set(topics)) - (1 if -1 in topics else 0)} (excl. noise -1)')
```

```python
# Save topic info
info = save_topic_info(model, '../outputs/analysis/topic_model_results/caption_topics.csv')
info.head(15)
```

```python
# Topic × culture heatmap (novel: replaces topic × persona)
culture_heat = topic_culture_proportions(df, topics)
fig, ax = plt.subplots(figsize=(14, 8))
sns.heatmap(culture_heat.head(15), annot=True, fmt='.2f', cmap='Reds', ax=ax)
ax.set_title('Topic × Culture Proportions (Qwen3.5-2B, column-normalised)')
plt.tight_layout()
plt.savefig('../outputs/figures/04_topic_culture_heatmap.pdf', bbox_inches='tight')
plt.show()
```
